## 2. Loading the Dataset

```python
import pandas as pd

# Load MovieLens datasets
movies = pd.read_csv("../data/raw/ml-latest-small/movies.csv")
ratings = pd.read_csv("../data/raw/ml-latest-small/ratings.csv")
tags = pd.read_csv("../data/raw/ml-latest-small/tags.csv")
links = pd.read_csv("../data/raw/ml-latest-small/links.csv")
```

```python
# Check dataset dimensions
print("Movies:", movies.shape)
print("Ratings:", ratings.shape)
print("Tags:", tags.shape)
print("Links:", links.shape)
```

## 3. Dataset Overview

This section examines the structure, column names, data types, and completeness of the four datasets.

```python
# Preview the first five rows of each dataset
display(movies.head())
display(ratings.head())
display(tags.head())
display(links.head())
```

```python
# Check columns, data types, and non-null counts
print("Movies Dataset:")
movies.info()

print("\nRatings Dataset:")
ratings.info()

print("\nTags Dataset:")
tags.info()

print("\nLinks Dataset:")
links.info()
```

## 4. Basic Statistics

This section calculates the number of unique users and movies, and examines the distribution of user ratings.

```python
# Count unique users and movies
print("Unique users:", ratings["userId"].nunique())
print("Unique rated movies:", ratings["movieId"].nunique())
print("Total movies:", movies["movieId"].nunique())

# Calculate the number of movies without ratings
moviesWithoutRatings = movies["movieId"].nunique() - ratings["movieId"].nunique()

print("Movies without ratings:", moviesWithoutRatings)
```

```python
# Calculate the total possible user-movie combinations
totalUsers = ratings["userId"].nunique()
totalMovies = movies["movieId"].nunique()
possibleRatings = totalUsers * totalMovies

# Calculate sparsity
actualRatings = ratings.shape[0]
sparsity = (1 - actualRatings / possibleRatings) * 100

print(f"Data sparsity: {sparsity:.2f}%")
```

```python
# Count the frequency of each rating
ratingCounts = ratings["rating"].value_counts().sort_index()
display(ratingCounts)

# Find the most frequent rating and its frequency
mostFrequentRating = ratingCounts.idxmax()
mostFrequentCount = ratingCounts.max()

print("Most frequent rating:", mostFrequentRating)
print("Frequency:", mostFrequentCount)
```

## 5. Data Quality

This section checks for missing values and duplicate records across the four datasets.

```python
# Check missing values in each dataset
print("Missing values in Movies:")
print(movies.isnull().sum())

print("\nMissing values in Ratings:")
print(ratings.isnull().sum())

print("\nMissing values in Tags:")
print(tags.isnull().sum())

print("\nMissing values in Links:")
print(links.isnull().sum())
```

```python
# Check exact duplicate rows
print("Movies duplicates:", movies.duplicated().sum())
print("Ratings duplicates:", ratings.duplicated().sum())
print("Tags duplicates:", tags.duplicated().sum())
print("Links duplicates:", links.duplicated().sum())

# Check repeated user/movie combinations
duplicateUserMovieCombination = ratings.duplicated(
    subset=["userId", "movieId"]
).sum()

print("Duplicate User/Movie combinations:", duplicateUserMovieCombination)
```

## 6. Genre Analysis

This section examines the frequency of individual movie genres. Since a movie can belong to multiple genres, the genres are separated before counting.

```python
# Split genres into individual lists
genresSplit = movies["genres"].str.split("|")

# Count each genre separately
genreCounts = genresSplit.explode().value_counts()

display(genreCounts)

# Display the top 2 most frequent genres
print("Top 2 most frequent genres:")
display(genreCounts.head(2))
```
